```python
with open('multiline_input.fasta', 'r') as f_in, open('output.fasta', 'w') as f_out:
    dna = ''
    for ln in f_in:
        ln = ln.strip()
        if ln[0] == '>':
            if dna:
                f_out.write(dna + '\n')
                dna = ''
            f_out.write(ln + '\n')
        else:
            dna += ln
    if dna:
        f_out.write(dna + '\n')
```
